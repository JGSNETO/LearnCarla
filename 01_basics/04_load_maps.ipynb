```python
import carla
```

```python
# Connect to the CARLA server
client = carla.Client("localhost", 2000)
client.set_timeout(10.0)
```

```python
# Map to load
map_name = "Town01"

print(f"Loading {map_name}...")
```

```python
# Load the map
world = client.load_world(map_name)

print(f"Map loaded: {world.get_map().name}")
```
