```python
import carla
```

```python
client = carla.Client("localhost", 2000)
client.set_timeout(20)
```

```python
world = client.get_world()
```

```python
weather = world.get_weather()
```

```python
weather.sun_altitude_angle = -10
```

```python
TimeOfDay = carla.WeatherParameters(
    precipitation=0.0,
    sun_altitude_angle= 90
)
```

```python
#world.set_weather(weather)
world.set_weather(TimeOfDay)
```

```python
print("=== TIME OF THE DAY ===")
print(f"Sun altitude: {weather.sun_altitude_angle}")
print(f"Sun azimuth: {weather.sun_azimuth_angle}")
```
