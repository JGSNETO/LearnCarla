```python
import carla
import random
```

```python
CARLA_LOCALHOST = 'localhost'
CARLA_TIMEOUT = 10.0
CARLA_PORT = 2000
TRAFIC_MANAGER_PORT = 8000
DISTANCE_TO_LEADING_VEHICLE = 1.0
GLOBAL_PERCENTAGE_SPEED_DIFFERENCE = 15.0
```

```python
def connect_to_carla() -> tuple [carla.Client, carla.World]:

    try:
        client = carla.Client(CARLA_LOCALHOST, CARLA_PORT)
        client.set_timeout(CARLA_TIMEOUT)
        world = client.get_world()
    except RuntimeError:
        print("Carla failed to create a valid client.")
    finally:
        print("### Carla client connection established ###")
        
    return client, world
        
```
