```python
import carla
```

```python
client = carla.Client("localhost", 2000)
client.set_timeout(10)
```

```python
world = client.get_world()
```

```python
vehicle_bp_library = world.get_blueprint_library().filter("vehicle.*")
```

```python
spawn_points = world.get_map().get_spawn_points()
```

```python
vehicle = world.try_spawn_actor(
    vehicle_bp_library[0],
    spawn_points[0]
)
```

```python
if vehicle is not None:
    print("Vehicle spawned succefully!\n")
    print(f"Actor ID: {vehicle_bp_library[0].id}\n")
    print(f"Spawned point: {spawn_points[0]}")

    location = vehicle.get_location()

    print(
        f"Vehicle Location: "
        f"x={location.x:.2f}, "
        f"y={location.y:.2f}, "
        f"z={location.z:.2f}, "

    )
else:
    print("Failed to spawn the vehicle!")
```
