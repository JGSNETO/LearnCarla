## Running CARLA Traffic Simulation

CARLA provides example scripts for generating traffic and manually controlling a vehicle.

### Terminal 1 — Generate Traffic

Open a terminal and navigate to the CARLA examples directory:

```powershell
cd C:\CARLA\PythonAPI\examples
py generate_traffic.py
```

This starts the traffic manager and spawns vehicles controlled by autopilot.

### Terminal 2 — Manual Vehicle Control

Open a **second terminal** and run:

```powershell
cd C:\CARLA\PythonAPI\examples
py manual_control.py
```

This opens the manual control interface, allowing you to drive a vehicle using the keyboard.

### Workflow

```text
Terminal 1
    │
    └── generate_traffic.py
            │
            ├── Spawns vehicles
            └── Autopilot / Traffic Manager
                        
Terminal 2
    │
    └── manual_control.py
            │
            └── Manually controlled vehicle
```

> **Note:** Start the CARLA simulator before running either script.
>
> ```powershell
> cd C:\CARLA
> .\CarlaUE4.exe
> ```
>
> The executable name may differ depending on the CARLA package version.

````python
"""
Launching CARLA and connecting the client.

The client object serves to maintain the client's connection to the server and has a number of functions for applying commands
and loading or exporting data. We can load an alternative map or reload the current one (resetting to initial state) using the client object.

The port can be chosen as any available port and is set to 2000 by default, you can also choose a host different from localhost by using a 
computer's IP address. This way, the CARLA server can be run on a networked machine, while the python client can be run from a personal computer. This is particularly useful for differentiating the GPU used for running the CARLA simulator and that used for neural network training, both of which can be highly demanding on graphics hardware.
"""
import carla
import random

client = carla.Client('localhost', 2000)
world = client.get_world()
````
